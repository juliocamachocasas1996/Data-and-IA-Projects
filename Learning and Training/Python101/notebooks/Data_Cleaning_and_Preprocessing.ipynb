# Data Cleaning and Preprocessing
## Essential skills for real-world data analysis

## 1. Import Libraries and Load Sample Data

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, MinMaxScaler

# Create a sample dataset with common issues
data = {
    'ID': [1, 2, 3, 4, 5, 6, 7, 8],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank', 'Grace', 'Henry'],
    'Age': [25, 30, np.nan, 35, 28, 32, np.nan, 29],
    'Salary': [50000, 60000, 55000, np.nan, 52000, 61000, 58000, 59000],
    'Department': ['Sales', 'IT', 'Sales', 'IT', 'HR', 'IT', 'HR', 'Sales']
}

df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)
```

## 2. Data Inspection

```python
# Basic info
print("DataFrame Info:")
print(df.info())

print("\nDataFrame Description:")
print(df.describe())

print("\nData Types:")
print(df.dtypes)

print("\nMemory Usage:")
print(df.memory_usage(deep=True))
```

```python
# Missing values detection
print("Missing values:")
print(df.isnull().sum())

print("\nMissing value percentage:")
print((df.isnull().sum() / len(df) * 100).round(2))
```

## 3. Handling Missing Values

```python
# Create a copy for demonstration
df_copy = df.copy()

# Drop rows with any missing values
df_dropped = df_copy.dropna()
print("After dropping NaN rows:")
print(df_dropped)
print(f"Rows reduced from {len(df_copy)} to {len(df_dropped)}")
```

```python
# Fill missing values with mean
df_filled_mean = df.copy()
df_filled_mean['Age'] = df_filled_mean['Age'].fillna(df_filled_mean['Age'].mean())
df_filled_mean['Salary'] = df_filled_mean['Salary'].fillna(df_filled_mean['Salary'].mean())

print("After filling with mean:")
print(df_filled_mean)
print(f"\nMissing values: {df_filled_mean.isnull().sum().sum()}")
```

```python
# Forward fill and backward fill
df_ffill = df.copy()
df_ffill['Age'] = df_ffill['Age'].fillna(method='ffill')
df_ffill['Salary'] = df_ffill['Salary'].fillna(method='bfill')

print("After forward/backward fill:")
print(df_ffill)
```

## 4. Handling Duplicates

```python
# Create data with duplicates
df_dup = pd.DataFrame({
    'ID': [1, 2, 2, 3, 3, 3],
    'Name': ['Alice', 'Bob', 'Bob', 'Charlie', 'Charlie', 'Charlie'],
    'Value': [100, 200, 200, 300, 300, 300]
})

print("Data with duplicates:")
print(df_dup)

print("\nDuplicate rows:")
print(df_dup.duplicated())

print("\nAfter removing duplicates:")
df_no_dup = df_dup.drop_duplicates()
print(df_no_dup)
```

## 5. Type Conversion

```python
# Type conversion examples
df_types = pd.DataFrame({
    'Age': ['25', '30', '35'],
    'Salary': ['50000', '60000', '55000'],
    'Active': ['Yes', 'No', 'Yes']
})

print("Original dtypes:")
print(df_types.dtypes)
print(df_types)

# Convert to numeric
df_types['Age'] = df_types['Age'].astype('int')
df_types['Salary'] = pd.to_numeric(df_types['Salary'])

print("\nAfter conversion:")
print(df_types.dtypes)
print(df_types)
```

## 6. Scaling and Normalization

```python
# Create sample numeric data
data_scale = pd.DataFrame({
    'Age': [25, 30, 35],
    'Salary': [50000, 60000, 55000],
    'Experience': [2, 5, 7]
})

print("Original data:")
print(data_scale)

# Standard scaling (z-score normalization)
scaler = StandardScaler()
scaled_std = pd.DataFrame(
    scaler.fit_transform(data_scale),
    columns=data_scale.columns
)

print("\nStandardized (Z-score):")
print(scaled_std)
print(f"Mean: {scaled_std.mean().values}")
print(f"Std: {scaled_std.std().values}")
```

```python
# Min-Max scaling (0-1 normalization)
scaler_mm = MinMaxScaler()
scaled_mm = pd.DataFrame(
    scaler_mm.fit_transform(data_scale),
    columns=data_scale.columns
)

print("Min-Max Scaled (0-1):")
print(scaled_mm)
print(f"Min: {scaled_mm.min().values}")
print(f"Max: {scaled_mm.max().values}")
```

## 7. Encoding Categorical Variables

```python
# Create categorical data
df_cat = pd.DataFrame({
    'Department': ['Sales', 'IT', 'HR', 'Sales', 'IT'],
    'Level': ['Junior', 'Senior', 'Junior', 'Senior', 'Junior']
})

print("Original data:")
print(df_cat)

# One-hot encoding
df_onehot = pd.get_dummies(df_cat, columns=['Department', 'Level'])
print("\nOne-hot encoded:")
print(df_onehot)
```

```python
# Ordinal encoding (mapping)
df_ordinal = df_cat.copy()
level_map = {'Junior': 1, 'Senior': 2}
dept_map = {'Sales': 1, 'IT': 2, 'HR': 3}

df_ordinal['Level'] = df_ordinal['Level'].map(level_map)
df_ordinal['Department'] = df_ordinal['Department'].map(dept_map)

print("Ordinal encoded:")
print(df_ordinal)
```

## 8. Date/Time Parsing and Feature Extraction

```python
# Create date data
df_dates = pd.DataFrame({
    'Date': ['2023-01-15', '2023-02-20', '2023-03-10'],
    'Amount': [100, 200, 150]
})

print("Original data:")
print(df_dates)
print(df_dates.dtypes)

# Parse dates
df_dates['Date'] = pd.to_datetime(df_dates['Date'])

# Extract date features
df_dates['Year'] = df_dates['Date'].dt.year
df_dates['Month'] = df_dates['Date'].dt.month
df_dates['Day'] = df_dates['Date'].dt.day
df_dates['DayOfWeek'] = df_dates['Date'].dt.dayofweek

print("\nAfter parsing and feature extraction:")
print(df_dates)
```
