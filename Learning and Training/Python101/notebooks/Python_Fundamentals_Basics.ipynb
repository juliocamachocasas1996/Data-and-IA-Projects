## 1. Environment & Imports
First, let's verify our Python environment and import necessary libraries.

## 2. Variables & Data Types
Understanding how Python stores and manages data.

```python
# Variables and assignment
name = "Alice"
age = 30
height = 5.8
is_student = False

print(f"Name: {name}, Type: {type(name)}")
print(f"Age: {age}, Type: {type(age)}")
print(f"Height: {height}, Type: {type(height)}")
print(f"Is Student: {is_student}, Type: {type(is_student)}")
```

```python
# Type conversions
x = "123"
x_int = int(x)
x_float = float(x)

print(f"String: {x}, type: {type(x)}")
print(f"Integer: {x_int}, type: {type(x_int)}")
print(f"Float: {x_float}, type: {type(x_float)}")
```

## 3. Collections: Lists, Tuples, Sets, and Dictionaries

```python
# Lists - mutable, ordered
fruits = ['apple', 'banana', 'orange']
print(f"List: {fruits}")
print(f"First element: {fruits[0]}")
fruits.append('grape')
print(f"After append: {fruits}")

# Tuples - immutable, ordered
coordinates = (10, 20)
print(f"\nTuple: {coordinates}")
# coordinates[0] = 30  # This would cause an error

# Sets - unordered, unique
unique_numbers = {1, 2, 3, 3, 2}
print(f"\nSet: {unique_numbers}")

# Dictionaries - key-value pairs
person = {'name': 'Bob', 'age': 25, 'city': 'NYC'}
print(f"\nDict: {person}")
print(f"Person's name: {person['name']}")
```

## 4. Control Flow: If/Elif/Else

```python
# Conditional statements
score = 85

if score >= 90:
    grade = 'A'
elif score >= 80:
    grade = 'B'
elif score >= 70:
    grade = 'C'
else:
    grade = 'F'

print(f"Score: {score}, Grade: {grade}")
```

## 5. Loops: For and While

```python
# For loop with range
print("For loop with range:")
for i in range(5):
    print(i, end=' ')
    
print("\n\nFor loop with list:")
colors = ['red', 'green', 'blue']
for color in colors:
    print(f"Color: {color}")

print("\nFor loop with enumerate:")
for index, color in enumerate(colors):
    print(f"Index {index}: {color}")
```

```python
# While loop
count = 0
while count < 5:
    print(f"Count: {count}")
    count += 1
```

## 6. Comprehensions
Concise and efficient way to create collections.

```python
# List comprehension
squares = [x**2 for x in range(10)]
print(f"Squares: {squares}")

# List comprehension with condition
even_squares = [x**2 for x in range(10) if x % 2 == 0]
print(f"Even Squares: {even_squares}")

# Dictionary comprehension
char_count = {char: word.count(char) for char in set(word) for word in ['hello']}
print(f"\nChar count in 'hello': {char_count}")

# Set comprehension
unique_lengths = {len(word) for word in ['apple', 'pie', 'banana', 'cat']}
print(f"Unique word lengths: {unique_lengths}")
```

## 7. Functions
Creating reusable code blocks.

```python
# Simple function
def greet(name):
    """Greet a person by name."""
    return f"Hello, {name}!"

print(greet("Alice"))

# Function with default arguments
def power(base, exponent=2):
    """Raise base to the power of exponent."""
    return base ** exponent

print(f"2^3 = {power(2, 3)}")
print(f"5^2 = {power(5)}")
```

```python
# *args and **kwargs
def print_info(*args, **kwargs):
    """Print positional and keyword arguments."""
    print(f"Args: {args}")
    print(f"Kwargs: {kwargs}")

print_info(1, 2, 3, name='Alice', age=30)

# Lambda functions
add = lambda x, y: x + y
print(f"\nAdd 5 + 3 = {add(5, 3)}")
```

```python
# Type hints (Python 3.5+)
def calculate_area(radius: float) -> float:
    """Calculate the area of a circle."""
    import math
    return math.pi * radius ** 2

area = calculate_area(5)
print(f"Area of circle with radius 5: {area:.2f}")
```

## 8. Error Handling
Managing exceptions gracefully.

```python
def safe_divide(a, b):
    """Safely divide two numbers."""
    try:
        result = a / b
        return result
    except ZeroDivisionError:
        print("Error: Cannot divide by zero!")
        return None
    except TypeError:
        print("Error: Invalid types for division!")
        return None
    finally:
        print("Division attempt complete.")

print(f"10 / 2 = {safe_divide(10, 2)}")
print(f"10 / 0 = {safe_divide(10, 0)}")
```

## 9. Simple Unit Tests & Assertions

```python
def multiply(a, b):
    """Multiply two numbers."""
    return a * b

# Assertions
assert multiply(2, 3) == 6, "Multiplication test failed"
assert multiply(-1, 5) == -5, "Negative multiplication test failed"
assert multiply(0, 100) == 0, "Multiplication by zero test failed"

print("All multiplication tests passed!")
```

```python
# Simple test function
def is_even(n):
    return n % 2 == 0

# Test cases
test_cases = [
    (2, True),
    (3, False),
    (0, True),
    (-4, True),
]

for num, expected in test_cases:
    result = is_even(num)
    assert result == expected, f"Test failed for {num}: expected {expected}, got {result}"
    print(f"✓ is_even({num}) = {result}")

print("\nAll tests passed!")
```
